# Number Statistics 🔢📊

## Task 5 - Level 1 - Day 5

This project analyzes a list of numbers using Python.

The program:

- Identifies even numbers
- Identifies odd numbers
- Finds the largest number
- Finds the smallest number

````python
# Store numerical data in a Python list

numbers = [
    12, 7, 25, 4, 18,
    31, 9, 42, 16, 5,
    28, 11, 36, 3, 20
]

print("Numbers:", numbers)
````

## 1. Identify Even Numbers

A number is even if it is completely divisible by 2.

The modulo operator `%` is used to check the remainder.

If:

```python
number % 2 == 0

````python

### Cell 4 — Code

# Find even numbers

even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print("Even Numbers:", even_numbers)
````

## 2. Identify Odd Numbers

A number is odd if dividing it by 2 leaves a remainder of 1.

The condition:

```python
number % 2 != 0

````python

### Cell 6 — Code

# Find odd numbers

odd_numbers = []

for number in numbers:
    if number % 2 != 0:
        odd_numbers.append(number)

print("Odd Numbers:", odd_numbers)
````

## 3. Find the Largest Number

The `max()` function is used to find the largest value in the list.

````python
# Find the largest number

largest_number = max(numbers)

print("Largest Number:", largest_number)
````

## 4. Find the Smallest Number

The `min()` function is used to find the smallest value in the list.

````python
# Find the smallest number

smallest_number = min(numbers)

print("Smallest Number:", smallest_number)
````

## 5. Final Number Statistics

The following code displays all the results together.

````python
# Display final results

print("===================================")
print("         NUMBER STATISTICS")
print("===================================")

print("Numbers:", numbers)
print("Even Numbers:", even_numbers)
print("Odd Numbers:", odd_numbers)
print("Largest Number:", largest_number)
print("Smallest Number:", smallest_number) 
````
